# RAG Experimentation

Load the KB, chunk it, embed, and inspect retrieval. Runs in mock mode by default.

```python
import os, sys
os.environ.setdefault('EMBEDDINGS_PROVIDER','mock')
os.environ.setdefault('VECTOR_STORE','memory')
sys.path.insert(0, '..')
```

```python
from src.retrieval.document_loader import load_documents
from src.retrieval.chunking import chunk_documents
docs = load_documents()
chunks = chunk_documents(docs)
print(f'{len(docs)} docs -> {len(chunks)} chunks')
for c in chunks[:3]:
    print(c['source'], '::', c['text'][:80].replace(chr(10),' '), '...')
```

```python
from src.retrieval.retriever import Retriever
r = Retriever(); print('indexed', r.build_index(), 'chunks'); print(r.describe())
```

```python
for q, cat in [('refund within 7 days','refund_request'), ('how do I cancel','subscription_cancellation'), ('forgot my password','login_account_access')]:
    hits = r.retrieve(q, category=cat)
    print(q, '->', [(h.source, round(h.score,3)) for h in hits])
```
